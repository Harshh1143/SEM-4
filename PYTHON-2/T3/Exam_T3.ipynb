373. Using the Open Weather Map API, generate a 3 hourly 5 days weather forecast for Ahmedabad with details like minimum
temperature, maximum temperature, wind speed, humidity, and weather description. Display this data in the form of a Pandas
data frame with the column names being date_time, min_temp, max_temp, wind_speed, humidity, and weather_description.

```python
import pandas as pd
url = (
    "https://api.openweathermap.org/data/2.5/forecast"
    "?q=Ahmedabad,IN"
    "&units=metric"
    f"&appid={API_KEY}"
)

# Fetch data
response = requests.get(url)
data = response.json()

# Create list to store forecast records
forecast_data = []

# Extract required details
for item in data["list"]:
    forecast_data.append({
        "date_time": item["dt_txt"],
        "min_temp": item["main"]["temp_min"],
        "max_temp": item["main"]["temp_max"],
        "wind_speed": item["wind"]["speed"],
        "humidity": item["main"]["humidity"],
        "weather_description": item["weather"][0]["description"]
    })

# Convert to Pandas DataFrame
df = pd.DataFrame(forecast_data)

# Display DataFrame
print(df)

# Optional: View first few rows
print("\nFirst 5 Forecast Records:")
print(df.head())
```

375. Write a program for web scrapping using BeautifulSoup to scrape the following details from the given link and make a data
frame using that scraped data from
the page in a given link.
Link: https://www.politifact.com/factchecks
You will find 30 news articles with fact checks on this page. You need to scrape the following details from all the articles and
store that in a data frame. Statement of News, Date of News, Source of News. 


```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

# URL of the webpage
url = "https://www.politifact.com/factchecks"

# Send HTTP request
headers = {
    "User-Agent": "Mozilla/5.0"
}
response = requests.get(url, headers=headers)

# Parse HTML content
soup = BeautifulSoup(response.content, "html.parser")

# Lists to store scraped data
statements = []
dates = []
sources = []

# Find all fact-check articles
articles = soup.find_all("article", class_="m-statement")

# Loop through each article
for article in articles:
    
    # Extract Statement of News
    statement = article.find("div", class_="m-statement__quote")
    if statement:
        statements.append(statement.get_text(strip=True))
    else:
        statements.append("N/A")
    
    # Extract Source of News
    source = article.find("a", class_="m-statement__name")
    if source:
        sources.append(source.get_text(strip=True))
    else:
        sources.append("N/A")
    
    # Extract Date of News
    date = article.find("div", class_="m-statement__desc")
    if date:
        # Example text: "stated on May 24, 2025 in..."
        dates.append(date.get_text(strip=True))
    else:
        dates.append("N/A")

# Create DataFrame
df = pd.DataFrame({
    "Statement of News": statements,
    "Date of News": dates,
    "Source of News": sources
})

# Display DataFrame
df
```

376. Write a program of web scrapping using BeautifulSoup to scrape the given data from the following link.
https://editorial.rottentomatoes.com/guide/popular-movies/
On the above link, you’ll find 30 Popular movies. Scrape the Movie Title and Rating of that particular movie and make a
Dataframe of the same.

```python
import requests 
from bs4 import BeautifulSoup
import pandas as pd 
url = 'https://editorial.rottentomatoes.com/guide/popular-movies/'
response = requests.get(url)
soup = BeautifulSoup(response.content,'html.parser')
m = soup.find_all('div',class_='content-body')[0]
df = {
    'title' : [],
    'rating' : []
}
for i in (soup.find_all('div',class_='block-countdown rkv-block')) :
    df['title'].append(i.find('div',class_='meta-title-wrapper').text.strip().split('\n')[0])
    df['rating'].append(i.find('div',class_='meta-scores-wrapper').text.strip())
df = pd.DataFrame(df)
df
```
