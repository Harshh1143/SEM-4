```python
soup = BeautifulSoup(res,'html.parser')
```

```python
# <div class="_4rR01T">Huawei P9 (Mystic Silver, 32 GB)</div>
df = {'mobilename':[],
    'star':[],
    'review':[],
    'rating':[],
    'camera':[],
    'new_price':[],
    'old_price':[],
    'discount':[]}

for i in (soup.find_all('div',class_='_3pLy-c')):
    df['mobilename'].append(i.find('div',class_='_4rR01T').text)
    df['star'].append(i.find('div',class_="_3LWZlK").text)
    x = (i.find('div',class_="gUuXy-").text)
    x = x.split('&')
    df['rating'].append(x[0].strip())
    df['review'].append(x[1].strip()) 
    df['camera'].append(i.find_all('li',class_="rgWa7D")[2].text)
    
    df['new_price'].append(i.find('div',class_='_30jeq3').text)
    
    try:
        df['old_price'].append(i.find('div',class_='_3I9_wc').text)
    except Exception:
        df['old_price'].append('Not Found')
    # print((i.find('div',class_='_3I9_wc').text))
    try:
        df['discount'].append(i.find('div',class_='_3Ay6Sb').text)
    except Exception:
        df['discount'].append('Not Found')
    # print((i.find('div',class_='_3Ay6Sb').text))

df1 = pd.DataFrame(df)
df1
```
