```python
df2 = pd.DataFrame(np.random.randint(1,100,(10,4)),columns=['A','B','C','D'])
df2
```

```python
df2["Total"]=df2.apply(sum,axis=1)
df2
```

```python
df2["Average"]=df2.apply(lambda x:(x["A"]+x["B"]+x["C"])/3,axis=1)
```

```python
df2
```

```python
df2["Results"]=df2.apply(lambda x: "Fail" if ((x["A"]<35)  or (x["B"]<35) or (x["C"]<35) or (x["D"]<35)) else "Pass",axis=1)
```

```python
df2
```

```python
def Grade(Total):
    if Total>200:
        return "A+"
    elif Total>180:
        return "A"
    elif Total>170:
        return "B"
    elif Total>150:
        return "C"
    return "FAIL"
```

```python
df2["Grade"]=df2["Total"].apply(Grade)
```
