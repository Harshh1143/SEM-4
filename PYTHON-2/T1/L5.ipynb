```python
items_path = "items.csv"
q1_path = "quarter-1.csv"
q2_path = "quarter-2.csv"
q3_path = "quarter-3.csv"

q1= pd.read_csv(q1_path)
q2 = pd.read_csv(q2_path)
q3 = pd.read_csv(q3_path)

items = pd.read_csv(items_path)
```

```python
d1=pd.concat((q1,q2,q3))
d1
```

```python
d2 = pd.concat((q1,q2,q3),ignore_index=False,keys=["Q1","Q2","Q3"])
d2
```

```python
items
```

```python
d2
```

```python
d2 = d2.reset_index()
```

```python
d2
```

```python
d2 = pd.merge(d2,items,on='item_id')
d2
```

```python
d2.groupby(['level_0','item_name'],as_index=False)['quantity'].sum().sort_values('quantity',ascending=False).drop_duplicates(subset=['level_0'])
```

```python
d2['price']=d2['item_price'].apply(lambda x : float(x[1:]))
```

```python
d2.price
```

```python
d2['totalprice']=d2['quantity']*d2['price']
```

```python
d2
```

```python
x = d2.groupby(['level_0','item_name'],as_index=False)['totalprice'].sum().sort_values('totalprice',ascending=False).drop_duplicates(subset=['level_0'])
```

```python
x.groupby('level_0')['totalprice'].mean()
```
