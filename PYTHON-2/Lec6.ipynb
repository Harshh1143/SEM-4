```python
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.io as pio
pio.renderers.default='%notebookk'
```

```python
df = pd.read_csv('social_media_engagement.csv')
```

```python
df
```

```python
df.dtypes
```

```python
fig = px.bar(data_frame=df,x='post_type',y='likes',text_auto=True,color='platform',color_discrete_sequence=['black','blue','green'],height=700,width=700)
fig.show()
```

```python
fig = px.bar(data_frame=df,x='post_type',y='likes',text_auto=True,
             color='comments',color_discrete_sequence=['black','blue','green'],
             height=700,width=700,
             hover_data=['post_day','sentiment_score']
            )
fig.show()
```

```python
px.bar(data_frame=df,x='post_type',y=['likes','comments','shares'],color='platform'
      ,text='comments',barmode='stack')
```

```python
px.bar(data_frame=df,x='post_type',y=['likes','comments','shares'],
       color='platform'
      ,text='comments',
       barmode='group',
      pattern_shape='sentiment_score')
```

```python
fig = px.bar(data_frame=df,x='post_type',y='likes',
             text_auto=True,color='platform',
             color_discrete_map={'Facebook':'black','Instagram':'blue','Twitter':'green'},
             title='platform wise score',
             height=1500,width=2000,
             facet_col='sentiment_score',
            facet_row='post_day')
fig.show()
```

```python
px.scatter(data_frame=df,
          x='likes',y='comments',
          color='platform',
          symbol='post_type')
```

```python
px.scatter(data_frame=df,
          x='likes',y='comments',
          size='shares')
```

```python
px.scatter(data_frame=df,
          x='likes',y='comments',
          size='shares',hover_data=['post_day'],facet_col='sentiment_score',facet_row='post_day')
```

```python
px.scatter_3d(data_frame=df,height=1000,width=700,x='likes',
             y='comments',z='shares',color='platform',size='likes')
```

```python
px.scatter_matrix(data_frame=df,height=1000,width=700,
              dimensions=['likes','comments','shares'],
              color='platform',size='likes')
```

```python
px.line(data_frame=df,x='post_id',y='likes',
       color='post_type',hover_data='sentiment_score',markers=True)
```

```python
data = df.groupby("platform",as_index=False)[['likes','comments','shares']].sum()
```

```python
data
```

```python
fig = px.pie(data_frame=data,names='platform',values='likes')
fig.update_traces(textinfo='percent+label',textposition='outside',pull=[0.2,0,0])
```
